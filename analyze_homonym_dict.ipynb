```python
import pandas as pd
```

```python
homonym_dict = pd.read_json("datasets_pre_defined/huggingface.jsonl",
    lines=True,
    encoding="utf-8")
```

```python
homonym_dict.shape
```

```python
homonym_dict = homonym_dict[
    homonym_dict["examples"].apply(
        lambda examples: isinstance(examples, list) and len(examples) > 0
    )
].copy()

print(f"Remaining records: {len(homonym_dict)}")
print(f"Remaining lemmas: {homonym_dict['lemma'].nunique()}")

homonym_dict["gloss_key"] = homonym_dict["gloss"].apply(
    lambda glosses: " | ".join(glosses) if isinstance(glosses, list) else str(glosses)
)

gloss_counts = homonym_dict.groupby("lemma")["gloss_key"].transform("nunique")

homonym_dict = homonym_dict[
    gloss_counts > 1
].drop(columns=["gloss_key"]).reset_index(drop=True)

print(f"Remaining records: {len(homonym_dict)}")
print(f"Remaining lemmas: {homonym_dict['lemma'].nunique()}")
```

```python
homonym_dict
```

```python
homonym_dict.to_json(
      "datasets_pre_defined/homonym_dictionary_clean.jsonl",
      orient="records",
      lines=True,
      force_ascii=False,
  )
```
